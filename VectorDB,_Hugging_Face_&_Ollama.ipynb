**1. What is a Vector Database (VectorDB) and how is it different from traditional databases?**

**Vector Database**

A Vector Database (VectorDB) is a database designed to store and search vector embeddings generated from machine learning models.

Embeddings convert data such as:

- text

- images

- audio

- documents

into numerical vectors.

These vectors capture the semantic meaning of data and allow similarity search.

VectorDBs are widely used in:

- AI search engines

- recommendation systems

- RAG applications

- semantic search

- chatbots

**Traditional Database**

Traditional databases store structured data in rows and columns.

Examples:

- MySQL

- PostgreSQL

- Oracle

- SQL Server

They perform queries using:

- exact matching

- filtering

- relational joins

### Key Differences

| Feature | Vector Database | Traditional Database |
|--------|----------------|---------------------|
| Data Type | Embeddings (vectors) | Structured data |
| Search Method | Similarity search | Exact matching |
| Query Type | Nearest neighbor search | SQL queries |
| AI Capability | Designed for AI applications | Limited AI support |
| Examples | ChromaDB, Pinecone, Weaviate | MySQL, PostgreSQL |

**Example**

Query:

"Find documents similar to AI in healthcare"

VectorDB will return:

- Medical AI research papers

- Healthcare machine learning studies

Even if the exact phrase does not exist.

**6. Install and set up Chroma DB, and insert sample vector data for semantic search.**

```python
!pip install chromadb sentence-transformers
```

```python
import chromadb
from sentence_transformers import SentenceTransformer

# Initialize embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Create Chroma client
client = chromadb.Client()

# Create collection
collection = client.create_collection(name="documents")

# Sample data
docs = [
    "AI is transforming healthcare",
    "Machine learning improves predictions",
    "Vector databases enable semantic search"
]

# Generate embeddings
embeddings = model.encode(docs)

# Insert into database
collection.add(
    documents=docs,
    embeddings=embeddings.tolist(),
    ids=["1","2","3"]
)

# Query
query = model.encode(["AI in medicine"])

results = collection.query(
    query_embeddings=query.tolist(),
    n_results=2
)

print(results)
```

**7. Demonstrate how to download and fine-tune a Hugging Face model for a text generation task.**

```python
!pip install transformers datasets accelerate
```

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer, Trainer, TrainingArguments, DataCollatorForLanguageModeling
from datasets import load_dataset

# Load dataset
dataset = load_dataset("wikitext", "wikitext-2-raw-v1")

# Load tokenizer and model
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

tokenizer.pad_token = tokenizer.eos_token

# Tokenization function
def tokenize_function(examples):
    return tokenizer(
        examples["text"],
        padding="max_length",
        truncation=True,
        max_length=128
    )

# Tokenize dataset
tokenized_dataset = dataset.map(tokenize_function, batched=True)

# Remove unused columns
tokenized_dataset = tokenized_dataset.remove_columns(["text"])

# Data collator
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

# Training arguments
training_args = TrainingArguments(
    output_dir="./results",
    per_device_train_batch_size=2,
    num_train_epochs=1,
    logging_steps=50
)

# Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset["train"],
    data_collator=data_collator
)

# Train model
trainer.train()
```

**9. Implement a basic RAG system using Ollama with Llama3.**

```python
!pip install langchain langchain-community langchain-core chromadb sentence-transformers
```

```python
from langchain_community.llms import Ollama
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
```

```python
# Import libraries
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings

# Sample documents
documents = [
    "Machine learning is a subset of artificial intelligence.",
    "Deep learning uses neural networks with many layers.",
    "Vector databases store embeddings for semantic search."
]

# Initialize embedding model
embedding = HuggingFaceEmbeddings()

# Create vector database
vectordb = Chroma.from_texts(documents, embedding)

# User query
query = "What is machine learning?"

# Retrieve relevant document
docs = vectordb.similarity_search(query)

print("Query:", query)
print("Retrieved Document:", docs[0].page_content)
```

```python
# Import libraries
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

documents = [
    "Diabetes is a chronic disease that occurs when blood sugar levels are too high.",
    "Common symptoms include increased thirst, frequent urination, and fatigue.",
    "Regular exercise and healthy diet help control diabetes."
]

# Initialize embedding model
embedding = HuggingFaceEmbeddings()

# Store documents in vector database
vectordb = Chroma.from_texts(documents, embedding)

# User question
query = "What are symptoms of diabetes?"

# Retrieve relevant information
docs = vectordb.similarity_search(query)

print("User Question:", query)
print("Retrieved Medical Information:", docs[0].page_content)
```
